```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_pickle('../data/raw/treadmill/FIRE_0_8_processed.pkl') #or 'FIRE_1_2_processed.pkl'
df.head()
```

```python
fig, axs = plt.subplots(3, 1, figsize=(10, 6), layout="constrained")
axs[0].plot(df.rel_time, df.Fx_R)
axs[0].plot(df.rel_time, df.Fy_R)
axs[0].plot(df.rel_time, df.Fz_R)
axs[0].legend(['M-L', 'A-P', 'V'], loc='lower left')
axs[0].set_ylabel('Force [N]')
axs[0].grid(axis='x')

axs[1].plot(df.rel_time, df.Fx_L)
axs[1].plot(df.rel_time, df.Fy_L)
axs[1].plot(df.rel_time, df.Fz_L)
axs[1].legend(['M-L', 'A-P', 'V'], loc='lower left')
axs[1].set_ylabel('Force [N]')
axs[1].grid(axis='x')

axs[2].plot(df.rel_time, df.speed)
axs[2].legend(['Set Speed'], loc='lower left')
axs[2].set_ylabel('Set Speed [m/s]')
axs[2].grid(axis='x')
```
